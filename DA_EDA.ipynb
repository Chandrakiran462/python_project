### EDA-6/DC-4 : Identify/Handling Outliers (Boxplot -IQR, Z-Score Clipping)


```python
# Interview question: Outliers exist in Numerical or Categorical data?
# Ans: Both
```

```python
sns.boxplot(x = 'marital-status', y = 'age', hue = 'sex', data = df)
plt.xticks(rotation = 45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df['age']
)

plt.title('Age Distribution and Outliers')
plt.xlabel('Age')

plt.tight_layout()
plt.show()
```

```python
# Finding which records are actually Outliers and How many?
Q1 = df.age.quantile(.25)
print(Q1)
```

```python
Q3 = df.age.quantile(.75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
print(lower_bound)

upper_bound = Q3 + 1.5 * IQR
print(upper_bound)
```

```python
# Find Outliers: < LB  | > UB

outliers = df[(df.age < lower_bound) | (df.age > upper_bound)]

outliers

```

```python
df1 = df.copy()
```

```python
# find the records where age > lower bound & age < upper bound

df1 = df1[(df1.age >= lower_bound) & (df1.age <= upper_bound)]
```

```python
df.shape
```

```python
df1.shape
```

### Replace the Outliers with Mean or Median

```python
df2 = df.copy()

median_value = df2.age.median()
median_value
```

### EDA-6/DC-4 : Identify/Handling Outliers using Z-Score Clipping


```python
# Def : How many standard deviation, data point is away from the mean.

# Here we are assuming that data is normal distribution.
# we should apply z-score method to find the outliers if data is normally distributed.
# if data point (age) z-score > 3 or z-score < -3, will be consider as outliers

# z-score = (data - mean) / SD
```

```python
# Finding the z-score based on age column

df3 = df.copy()

df3['z-score'] = (df3.age - df3.age.mean()) / df3.age.std()

df3
```

```python
df3[(df3['z-score'] < -3 ) | (df3['z-score'] > 3)]
```

```python
df3 = df3[(df3['z-score'] >= -3 ) & (df3['z-score'] <= 3)]
df3
```

```python
# Replace outliers with mean value

mean_value = int(df3.age.mean())
mean_value
```

```python
df3 = df.copy()

df3['z-score'] = (df3.age - df3.age.mean()) / df3.age.std()

df3.loc[(df3['z-score'] < -3 ) | (df3['z-score'] > 3), 'age'] = mean_value
```

```python
df3
```

```python
df3.age.unique()
```

### EDA-6 : Identify / Handle Outliers using Clipping technique

```python
df1 = df.copy()
```

```python
# Clipping is replacing based on lower_bound and upper_bound

df1.age = df1.age.clip(lower = lower_bound, upper = upper_bound)
```

### Clipping Using Winsorization

```python
# Winsorization is a clipping technique which use the percentage fir clipping
```

```python
from scipy.stats.mstats import winsorize
```

```python
df2 = df.copy()

df2.age = winsorize(df2.age, limits = (0.05, 0.10))
```
