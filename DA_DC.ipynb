## DC-2 : Handling the NULL/NaN values 

```python
# Approaches to handle the NULL/NaN values

# Approach-1 : Delete the record or column which is having the NULL values.
       # 1.1 : When number of NULL values are very very less as compare to dataset size.
       # 1.2 : if number of NULL values in a column > 70% & column is not impacting your final analysis.

# Approach-2 : Fill the Constant value in place of NULL.

# Approach-3 : Fill the Mean/Median/Mode values in place of NULL value.

# Interview Question: When to use Mean/Median/Mode to replace Null Value?
# When to use Mean:
    # 1. Data must be Numerical.
    # 2. When the data is Normally Distributed.
    # 3. When we don't have the Outliers.
# When to use Median: 
    # 1. When data having outliers.
    # 2. When data is NOT Normally Distributed.
# When to use Mode: 
    # 1. When we have Categorical data or Discrete data (repeated) with frequency.

# Approach-4 : bfill() & ffill() functions

# Approach-5 : Interpolate() function

# Approach-6 : Supervised ML model
```

### Approach-4 : bfill() & ffill() functions

### Approach-5 : Interpolate() function


```python
# Def : Interpolate function estimate the missing/null values in a contineous Series.
# Types of Interpolate() functions:
    # 1. Linear : if data is having linear releationship
         # Linear cannou handle the starting NaN vlaue
         # Linear will put the last valid value in the last NaN value
    # 2. Polynomial : if you have polynomial releationship in data
         # Polynomial don't insert the last NaN values
         # Polynomial also cannot handle the stating NaN value
    # 3. Spline : smooth curve
    # 4. Time : When we have time-series data




# Small Example

Number_GF = [np.nan, 1, 2, 3, 4, 5, np.nan, np.nan, 8, 9, np.nan, np.nan]

df4 = pd.DataFrame(Number_GF)

df4.interpolate(method = 'linear', order = 2, inplace = True)
df4
```

```python
Number_GF = [np.nan, 1, 2, 3, 4, 5, np.nan, np.nan, 8, 9, np.nan, np.nan]

df4 = pd.DataFrame(Number_GF)

df4.interpolate(method = 'polynomial', order = 2, inplace = True)
df4
```

```python
# Finding which records are actually Outliers and How many?
Q1 = df.age.quantile(.25)
print(Q1)
```

```python
Q3 = df.age.quantile(.75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
print(lower_bound)

upper_bound = Q3 + 1.5 * IQR
print(upper_bound)
```

```python
# Find Outliers: < LB  | > UB

outliers = df[(df.age < lower_bound) | (df.age > upper_bound)]

outliers

```

### DC-4 : Handling Outliers using Z-Score Clipping


```python
# Def : How many standard deviation, data point is away from the mean.

# Here we are assuming that data is normal distribution.
# we should apply z-score method to find the outliers if data is normally distributed.
# if data point (age) z-score > 3 or z-score < -3, will be consider as outliers

# z-score = (data - mean) / SD
```

```python
# Finding the z-score based on age column

df3 = df.copy()

df3['z-score'] = (df3.age - df3.age.mean()) / df3.age.std()

df3
```

```python
df3[(df3['z-score'] < -3 ) | (df3['z-score'] > 3)]
```

```python
df3 = df3[(df3['z-score'] >= -3 ) & (df3['z-score'] <= 3)]
df3
```

```python
# Replace outliers with mean value

mean_value = int(df3.age.mean())
mean_value
```

```python
df3 = df.copy()

df3['z-score'] = (df3.age - df3.age.mean()) / df3.age.std()

df3.loc[(df3['z-score'] < -3 ) | (df3['z-score'] > 3), 'age'] = mean_value
```
